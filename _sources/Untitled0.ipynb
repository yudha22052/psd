```python
# Konversi kolom 'tanggal' ke tipe datetime
data['Date and Month'] = pd.to_datetime(data['Date and Month'])

# Tambahkan kolom X sebagai indeks waktu (contohnya, 0, 1, 2, ...)
data['X'] = range(len(data))

# Salin jumlah impor ke kolom Y
data['Y'] = data['Crude oil imports from the World to the US (a thousand barrels)']

# Menampilkan data setelah preprocessing
print("Data setelah preprocessing:")
print(data.head())
```

```python
# Hitung kolom tambahan untuk regresi linear
data['XY'] = data['X'] * data['Y']  # Perkalian X dan Y
data['XX'] = data['X'] ** 2         # Kuadrat dari X

# Menampilkan data dengan kolom tambahan
print("Data dengan kolom XY dan XX:")
print(data[['X', 'Y', 'XY', 'XX']].head())
```

```python
# Hitung nilai a dan b
n = len(data)
sum_X = data['X'].sum()
sum_Y = data['Y'].sum()
sum_XY = data['XY'].sum()
sum_XX = data['XX'].sum()

b = (n * sum_XY - sum_X * sum_Y) / (n * sum_XX - sum_X**2)
a = (sum_Y - b * sum_X) / n

# Cetak hasil a dan b
print(f"Nilai a (intersep): {a:.2f}")
print(f"Nilai b (slope): {b:.2f}")
```
